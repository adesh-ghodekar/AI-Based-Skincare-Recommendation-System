```python
import sqlite3

conn = sqlite3.connect('progress.db')  # Update to your db file
cursor = conn.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS progress (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    username TEXT,
    image_path TEXT,
    skin_type TEXT,
    conditions TEXT,
    routine_json TEXT,
    notes TEXT,
    date TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
''')

conn.commit()
conn.close()
print("✅ Progress table created successfully.")
```
